##**6. Understanding Model Evaluation Metrics**
Three important evaluation metrics are used to compare forecasting performance.

A. Mean Absolute Error (MAE)

MAE measures the average absolute difference between actual sales and predicted sales.

A lower MAE indicates that the model produces predictions closer to actual sales.

Formula:

MAE = (1/n) × Σ |Actual Sales − Predicted Sales|

B. Root Mean Squared Error (RMSE)

RMSE measures prediction errors while giving greater importance to larger errors.

A lower RMSE indicates better forecasting performance, particularly when large prediction mistakes are important.

Formula:

RMSE = √[(1/n) × Σ(Actual Sales − Predicted Sales)²]

C. R² Score

R² measures how much variation in actual sales is explained by the forecasting model.

A higher R² generally indicates better explanatory performance.

**For example:**
- 0.86 = 86%
- 0.93 = 93%
- 0.94 = 94%

An R² score of 0.94 means the model explains approximately 94% of the variation in the evaluated dataset. It does not mean that 94% of individual predictions are correct.

```python
# ============================================================
# INTEGRATED RETAIL ANALYTICS FOR STORE OPTIMIZATION
# AND DEMAND FORECASTING
# ============================================================

# 1. Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import IsolationForest
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from statsmodels.tsa.statespace.sarimax import SARIMAX


# ============================================================
# 2. DATA LOADING AND PREPROCESSING
# ============================================================

print("\n1. DATA LOADING AND PREPROCESSING")

# Load datasets

df_sales = pd.read_csv("data/train.csv")
df_stores = pd.read_csv("data/stores.csv")
df_features = pd.read_csv("data/features.csv")

# Convert Date columns into datetime format

df_sales["Date"] = pd.to_datetime(df_sales["Date"])
df_features["Date"] = pd.to_datetime(df_features["Date"])

# Merge sales and features datasets

df = pd.merge(
    df_sales,
    df_features,
    on=["Store", "Date", "IsHoliday"],
    how="left"
)

# Merge store information

df = pd.merge(
    df,
    df_stores,
    on="Store",
    how="left"
)

# Handle missing Markdown values

markdown_cols = [
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

for col in markdown_cols:
    if col in df.columns:
        df[col] = df[col].fillna(0)

# Handle missing numerical values

numerical_cols = df.select_dtypes(
    include=["number"]
).columns

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())

# Handle missing categorical values

categorical_cols = df.select_dtypes(
    include=["object"]
).columns

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")

# Create time-based features

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)
df["Quarter"] = df["Date"].dt.quarter

# Sort records by date

df = df.sort_values("Date").reset_index(drop=True)

print("Dataset merged successfully!")
print("Dataset Shape:", df.shape)

print(df.head())


# ============================================================
# 3. ANOMALY DETECTION USING ISOLATION FOREST
# ============================================================

print("\n2. ANOMALY DETECTION")

# Select sales and markdown features

anomaly_features = [
    "Weekly_Sales",
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

# Keep only available columns

anomaly_features = [
    col for col in anomaly_features
    if col in df.columns
]

# Prepare input data

X_anomaly = df[anomaly_features].copy()

# Train Isolation Forest

iso = IsolationForest(
    contamination=0.01,
    random_state=42
)

df["Anomaly"] = iso.fit_predict(X_anomaly)

# Convert results into readable labels

df["Anomaly_Label"] = df["Anomaly"].map({
    1: "Normal",
    -1: "Anomaly"
})

# Display anomaly results

print(df["Anomaly_Label"].value_counts())

print("\nAnomaly Records:")
print(
    df[df["Anomaly"] == -1][
        ["Store", "Dept", "Date", "Weekly_Sales"]
    ].head(10)
)

# Visualize anomalies

plt.figure(figsize=(12, 5))

sns.scatterplot(
    data=df,
    x="Date",
    y="Weekly_Sales",
    hue="Anomaly_Label",
    alpha=0.6
)

plt.title("Retail Sales Anomaly Detection")
plt.xlabel("Date")
plt.ylabel("Weekly Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 4. STORE SEGMENTATION USING K-MEANS
# ============================================================

print("\n3. STORE SEGMENTATION")

# Calculate store-level performance

store_features = df.groupby("Store").agg({

    "Weekly_Sales": "mean",
    "MarkDown1": "mean",
    "Size": "mean"

}).reset_index()

# Add other markdown columns

for col in [
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]:
    if col in df.columns:
        markdown_avg = df.groupby("Store")[col].mean().reset_index()
        store_features = store_features.merge(
            markdown_avg,
            on="Store",
            how="left"
        )

# Select features for clustering

cluster_cols = [
    col for col in store_features.columns
    if col != "Store"
]

X_cluster = store_features[cluster_cols].copy()

# Scale features

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_cluster)

# Apply K-Means clustering

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

store_features["Store_Segment"] = kmeans.fit_predict(
    X_scaled
)

# Display store segments

print("\nStore Segmentation Results:")

print(
    store_features[
        ["Store", "Weekly_Sales", "Store_Segment"]
    ]
)

# Visualize store segmentation

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=store_features,
    x="Size",
    y="Weekly_Sales",
    hue="Store_Segment",
    palette="viridis",
    s=100
)

plt.title("Store Segmentation Using K-Means")
plt.xlabel("Store Size")
plt.ylabel("Average Weekly Sales")
plt.tight_layout()
plt.show()


# ============================================================
# 5. DEMAND FORECASTING USING RANDOM FOREST
# ============================================================

print("\n4. DEMAND FORECASTING")

# Select input features

features = [
    "Store",
    "Dept",
    "Temperature",
    "Fuel_Price",
    "CPI",
    "Unemployment",
    "Size",
    "Year",
    "Month",
    "Week",
    "Quarter",
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5",
    "IsHoliday"
]

# Keep available features

features = [
    col for col in features
    if col in df.columns
]

# Prepare feature matrix and target

X = df[features].copy()
y = df["Weekly_Sales"]

# Convert categorical variables into numerical columns

X = pd.get_dummies(
    X,
    drop_first=True
)

# Fill any remaining missing values

X = X.fillna(0)

# Chronological train-test split

split_date = df["Date"].quantile(0.8)

train_mask = df["Date"] <= split_date
test_mask = df["Date"] > split_date

X_train = X.loc[train_mask]
X_test = X.loc[test_mask]

y_train = y.loc[train_mask]
y_test = y.loc[test_mask]

# Train Random Forest model

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

# Predict sales

y_pred = rf.predict(X_test)

# Evaluate model

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

print("\nRandom Forest Model Performance")

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))

# Compare actual and predicted sales

plt.figure(figsize=(10, 5))

plt.plot(
    y_test.values[:100],
    label="Actual Sales"
)

plt.plot(
    y_pred[:100],
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Weekly Sales")
plt.xlabel("Test Records")
plt.ylabel("Weekly Sales")
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 6. TIME-SERIES FORECASTING USING SARIMA
# ============================================================

print("\n5. TIME-SERIES FORECASTING")

# Aggregate total weekly sales by date

weekly_sales = (
    df.groupby("Date")["Weekly_Sales"]
    .sum()
    .sort_index()
)

# Check the available number of weeks

if len(weekly_sales) >= 60:

    # Train-test split

    split_index = int(len(weekly_sales) * 0.8)

    train_ts = weekly_sales.iloc[:split_index]
    test_ts = weekly_sales.iloc[split_index:]

    # Build SARIMA model

    sarima_model = SARIMAX(
        train_ts,
        order=(1, 1, 1),
        seasonal_order=(1, 1, 1, 52),
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    # Fit model

    sarima_result = sarima_model.fit(disp=False)

    # Forecast test period

    forecast = sarima_result.forecast(
        steps=len(test_ts)
    )

    # Evaluate forecast

    sarima_mae = mean_absolute_error(
        test_ts,
        forecast
    )

    sarima_rmse = np.sqrt(
        mean_squared_error(test_ts, forecast)
    )

    sarima_r2 = r2_score(
        test_ts,
        forecast
    )

    print("\nSARIMA Model Performance")

    print("MAE:", round(sarima_mae, 2))
    print("RMSE:", round(sarima_rmse, 2))
    print("R2 Score:", round(sarima_r2, 4))

    # Plot actual and forecast sales

    plt.figure(figsize=(12, 5))

    plt.plot(
        train_ts.index,
        train_ts,
        label="Training Sales"
    )

    plt.plot(
        test_ts.index,
        test_ts,
        label="Actual Sales"
    )

    plt.plot(
        forecast.index,
        forecast,
        label="Forecast Sales"
    )

    plt.title("SARIMA Weekly Sales Forecast")
    plt.xlabel("Date")
    plt.ylabel("Total Weekly Sales")
    plt.legend()
    plt.tight_layout()
    plt.show()

else:
    print(
        "Not enough weekly observations for "
        "the selected SARIMA seasonal model."
    )


# ============================================================
# 7. DEPARTMENT ASSOCIATION USING SALES CORRELATION
# ============================================================

print("\n6. DEPARTMENT ASSOCIATION ANALYSIS")

# Aggregate department sales by date

dept_pivot = df.pivot_table(
    index="Date",
    columns="Dept",
    values="Weekly_Sales",
    aggfunc="sum"
)

# Calculate department sales correlation

dept_corr = dept_pivot.corr()

print("\nDepartment Sales Correlation:")

print(dept_corr)

# Visualize correlation

plt.figure(figsize=(12, 8))

sns.heatmap(
    dept_corr,
    cmap="coolwarm",
    center=0
)

plt.title("Department Sales Correlation Heatmap")
plt.tight_layout()
plt.show()


# ============================================================
# 8. SAVE PROJECT OUTPUTS
# ============================================================

df.to_csv(
    "retail_analysis_results.csv",
    index=False
)

store_features.to_csv(
    "store_segmentation_results.csv",
    index=False
)

dept_corr.to_csv(
    "department_correlation_results.csv"
)

print("\nPROJECT COMPLETED SUCCESSFULLY!")
print("All result files have been saved.")
```

##**2. Python Code to Generate All Five Graphs**

The following code creates five separate charts using your actual merged dataset. It assumes that you have already completed the data-loading and preprocessing steps from the previous section.

Run this code after preparing your df DataFrame.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import IsolationForest
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Make sure the date is in datetime format
df["Date"] = pd.to_datetime(df["Date"])

sns.set_theme(style="whitegrid")


# =================================================
# GRAPH 1: WEEKLY SALES TREND AND SEASONALITY
# =================================================

weekly_trend = (
    df.groupby("Date")["Weekly_Sales"]
    .sum()
    .sort_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    weekly_trend.index,
    weekly_trend.values,
    linewidth=1.5,
    label="Weekly Sales"
)

plt.title("Weekly Sales Trend and Seasonality")
plt.xlabel("Year")
plt.ylabel("Total Weekly Sales")
plt.legend()
plt.tight_layout()
plt.show()


# =================================================
# GRAPH 2: ANOMALY DETECTION
# =================================================

anomaly_features = [
    "Weekly_Sales",
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

anomaly_features = [
    col for col in anomaly_features
    if col in df.columns
]

X_anomaly = df[anomaly_features].fillna(0)

iso = IsolationForest(
    contamination=0.01,
    random_state=42
)

df["Anomaly"] = iso.fit_predict(X_anomaly)

df["Anomaly_Label"] = df["Anomaly"].map({
    1: "Normal",
    -1: "Anomaly"
})

plt.figure(figsize=(12, 5))

sns.scatterplot(
    data=df,
    x="Date",
    y="Weekly_Sales",
    hue="Anomaly_Label",
    alpha=0.6
)

plt.title("Anomaly Detection in Retail Sales")
plt.xlabel("Date")
plt.ylabel("Weekly Sales")
plt.legend(title="Sales Category")
plt.tight_layout()
plt.show()


# =================================================
# GRAPH 3: STORE SEGMENTATION USING K-MEANS
# =================================================

store_features = df.groupby("Store").agg({
    "Weekly_Sales": "mean",
    "Size": "mean"
}).reset_index()

markdown_means = df.groupby("Store")[
    anomaly_features[1:]
].mean().reset_index()

store_features = store_features.merge(
    markdown_means,
    on="Store",
    how="left"
)

cluster_cols = [
    col for col in store_features.columns
    if col != "Store"
]

X_cluster = store_features[cluster_cols].fillna(0)

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_cluster)

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

store_features["Segment"] = kmeans.fit_predict(
    X_scaled
)

# PCA reduces the cluster features to two dimensions
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

pca_data = pca.fit_transform(X_scaled)

store_features["PCA1"] = pca_data[:, 0]
store_features["PCA2"] = pca_data[:, 1]

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=store_features,
    x="PCA1",
    y="PCA2",
    hue="Segment",
    palette="viridis",
    s=100
)

plt.title("Store Segmentation Using K-Means and PCA")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.legend(title="Store Segment")
plt.tight_layout()
plt.show()


# =================================================
# GRAPH 4: DEPARTMENT SALES CORRELATION
# =================================================

dept_pivot = df.pivot_table(
    index="Date",
    columns="Dept",
    values="Weekly_Sales",
    aggfunc="sum"
)

dept_corr = dept_pivot.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    dept_corr,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)

plt.title("Department Sales Correlation")
plt.tight_layout()
plt.show()


# =================================================
# GRAPH 5: ACTUAL VS PREDICTED SALES
# =================================================

features = [
    "Store",
    "Dept",
    "Temperature",
    "Fuel_Price",
    "CPI",
    "Unemployment",
    "Size",
    "Year",
    "Month",
    "Week",
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5",
    "IsHoliday"
]

features = [
    col for col in features
    if col in df.columns
]

X = pd.get_dummies(
    df[features],
    drop_first=True
)

X = X.fillna(0)

y = df["Weekly_Sales"]

# Chronological train-test split

df_sorted = df.sort_values("Date")

X = X.loc[df_sorted.index]
y = y.loc[df_sorted.index]

split = int(len(df_sorted) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)

# Evaluation metrics

print("MAE:", mean_absolute_error(y_test, y_pred))

print(
    "RMSE:",
    np.sqrt(mean_squared_error(y_test, y_pred))
)

print("R2 Score:", r2_score(y_test, y_pred))

# Plot actual and predicted sales

plt.figure(figsize=(12, 5))

plt.plot(
    y_test.values[:100],
    label="Actual Sales"
)

plt.plot(
    y_pred[:100],
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Sales - Random Forest")
plt.xlabel("Test Records")
plt.ylabel("Weekly Sales")
plt.legend()
plt.tight_layout()
plt.show()
```
